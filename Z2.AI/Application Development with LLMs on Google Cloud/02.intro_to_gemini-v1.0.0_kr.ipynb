더 많은 기능을 제공하는 도구도 많지만, 실제 애플리케이션에서는 아주 단순한 통합만 필요한 경우가 많습니다.

```python
PROJECT_ID = "YOUR_PROJECT_ID"  # @param {type:"string"}
REGION = "us-central1"  # @param {type:"string"}
```

이제 GenAI 라이브러리와 ipython 헬퍼를 임포트합니다. 생성형 모델은 보통 마크다운으로 응답하기 때문에 결과를 깔끔하게 표시할 수 있고, 목록이나 표도 아주 간단하게 표현할 수 있습니다. 원격 모델을 쓰려면 프로젝트 ID와 리전을 설정하고 사용할 모델을 골라야 합니다. 이 실습의 대부분은 채팅 형태로 진행되므로 지연 시간이 낮은 모델을 사용합니다. 더 빠르고 저렴하지만 지식의 폭은 가장 넓지 않습니다. 실습을 진행하면서 여러 그라운딩 기법으로 모델의 지식을 보강할 것이므로 모델 자체가 아주 박식할 필요는 없습니다.


```python
from google import genai
from IPython.display import display, Markdown, clear_output

client = genai.Client(
    vertexai=True, project=PROJECT_ID, location=REGION
)

MODEL_ID='gemini-2.5-pro'

chat_history= []

def chat(user_input):
    global chat_history


    user_message = genai.types.Content(
        role="user",
        parts=[genai.types.Part(text=user_input)]
    )

    chat_history.append(user_message)

    response = client.models.generate_content(
        model=MODEL_ID,
        contents = chat_history
    )

    # 결과에서 모델의 응답을 추출합니다.
    model_response_content = response.candidates[0].content
    model_text = model_response_content.parts[0].text

    display(Markdown(f"**Gemini:** {model_text}"))

    # 모델의 응답을 대화 기록 리스트에 덧붙입니다.
    chat_history.append(model_response_content)
```

```python
chat("안녕")
```

모델의 지식 컷오프와 한계를 확인해 봅시다. 기본 상태의 모델은 메모리가 없고, 학습 컷오프 시점 이후의 정보에도 접근할 수 없습니다.

## 그라운딩을 위한 시스템 메시지

대화 기록을 비우고, 모델에 전달되는 첫 부분으로 시스템 메시지가 들어가도록 chat 함수를 다시 작성해 봅시다. 호출할 때마다 모델을 지정한다는 점에 주목하세요. 즉, 대화 도중에도 모델을 바꿔서 추론 능력을 높이거나 지연 시간을 줄일 수 있습니다. 운영 환경에서 흔히 하는 일은 아니지만 중요한 특성입니다. 언어 모델 자체에는 메모리가 없어 매 호출마다 전체 대화가 전달되므로, 서로 다른 모델을 섞어 쓰는 일이 아주 쉽습니다.

GenAI의 `types` 라이브러리를 사용하면 각 구성 요소를 설정할 수 있습니다.

```python
from google.genai import types

chat_history= []

def chat(user_input):
    global chat_history

    user_message = genai.types.Content(
        role="user",
        parts=[genai.types.Part(text=user_input)]
    )

    chat_history.append(user_message)

    system_instruction = "당신은 날씨 봇입니다. 워싱턴의 날씨는 맑고, 영국의 날씨는 비가 옵니다."

    response = client.models.generate_content(
        model=MODEL_ID,
        contents=chat_history,
        config=types.GenerateContentConfig(
            system_instruction=system_instruction,
            max_output_tokens=10000,  # 출력 토큰 수 지정
            temperature=0.3, # temperature 설정
        ),
    )

    # 결과에서 모델의 응답을 추출합니다.
    model_response_content = response.candidates[0].content
    model_text = model_response_content.parts[0].text

    display(Markdown(f"**Gemini:** {model_text}"))

    # 모델의 응답을 대화 기록 리스트에 덧붙입니다.
    chat_history.append(model_response_content)
```

```python
chat("영국 날씨는 어때?")
```

```python
chat("워싱턴 날씨는 어때?")
```

현재 세대 모델의 정말 강력한 특징 중 하나는 여러 언어를 다룰 수 있다는 점입니다. 이는 원래 트랜스포머 설계와 최초의 모델인 `BERT`가 번역 작업을 처리하도록 설계되었기 때문입니다.

```python
chat("워싱턴 날씨는 어때? 프랑스어로 답변해 줘.")
```

## 도구: Google 검색

## 도구: Google 지도

```python
from google import genai
from google.genai import types
from IPython.display import display, Markdown

client = genai.Client(
    vertexai=True,
    project=PROJECT_ID,
    location="us-central1",  # @param {type:"string"}
)

prompt = "여기서 걸어서 15분 이내 거리에 있는 최고의 이탈리안 레스토랑은 어디인가요? 이름, 평점, 주소를 알려 주세요"

response = client.models.generate_content(
    model='gemini-2.5-pro',
    contents=prompt,
    config=types.GenerateContentConfig(
        tools=[types.Tool(google_maps=types.GoogleMaps())],
        # 필요하면 관련 위치 컨텍스트를 함께 제공합니다 (원본 주석에는 로스앤젤레스라고 되어 있지만, 아래 좌표는 영국 메이든헤드입니다)
        tool_config=types.ToolConfig(retrieval_config=types.RetrievalConfig(
            lat_lng=types.LatLng(
                latitude=51.5220755, longitude=-0.7177022))),
    ),
)

display(Markdown(response.text))

if grounding := response.candidates[0].grounding_metadata:
  if grounding.grounding_chunks:
    print("Google 지도 출처:")
    for chunk in grounding.grounding_chunks:
      display(Markdown(f'- [{chunk.maps.title}]({chunk.maps.uri})'))
```

## 검색

많은 경우 데이터 소스는 신뢰할 수 있는 데이터에 그라운딩되어야 합니다. 컨텍스트 윈도우가 100만 토큰이라 해도, 대규모 조직에서 모델이 올바르게 답하는 데 필요한 모든 정보를 담기에는 대개 부족합니다. 그래서 맥락을 인지하는 검색 기능을 구축해야 합니다. 핵심 개념은 큰 문서를 청크로 나누고, 각 청크를 수치로 표현해 사용자의 질의와 어떤 청크가 비슷한지 알아내는 것입니다. 이것이 검색 증강 생성(RAG)의 기반이며, 임베딩 모델을 사용해 데이터(주로 텍스트)를 벡터로 변환합니다.

여기서는 다국어 임베딩 모델인 최신 gemini-embedding 모델을 사용합니다. 이전 버전은 영어 전용 또는 다국어 버전으로 나뉘어 있었습니다. 다국어 모델의 장점은 하나의 도구로 서로 다른 언어의 문서를 넘나들며 검색할 수 있다는 점입니다.

비슷한 단어 두 쌍의 임베딩을 구해 봅시다. 벡터 유사도를 계산할 헬퍼 함수도 필요합니다.

아래 수식들은 $n$차원 공간의 두 벡터 $\mathbf{A}$와 $\mathbf{B}$ 사이의 유사도 또는 거리를 나타냅니다. 여기서 $\mathbf{A} = (a_1, a_2, \dots, a_n)$, $\mathbf{B} = (b_1, b_2, \dots, b_n)$입니다.

#### L1 거리 (맨해튼 거리)

L1 거리는 두 벡터의 각 성분 차이의 절댓값을 모두 더한 값입니다. 도시의 격자형 도로를 택시가 이동하는 거리와 같다고 해서 맨해튼 거리 또는 택시 거리라고도 부릅니다. 값이 작을수록 가깝습니다(좋습니다).

$$D_{L1}(\mathbf{A}, \mathbf{B}) = \sum_{i=1}^{n} |a_i - b_i|$$

#### L2 거리 (유클리드 거리)

L2 거리는 유클리드 공간에서 두 점 사이의 일반적인 직선 거리입니다. 피타고라스 정리에 따라 각 성분 차이의 제곱을 모두 더한 뒤 제곱근을 취한 값입니다. 값이 작을수록 가깝습니다(좋습니다).

$$D_{L2}(\mathbf{A}, \mathbf{B}) = \sqrt{\sum_{i=1}^{n} (a_i - b_i)^2}$$

#### 코사인 유사도

코사인 유사도는 두 벡터가 이루는 각 $\theta$의 코사인 값을 측정합니다. 벡터의 크기와 무관하게 방향이 얼마나 비슷한지를 나타냅니다. 방향이 같으면 1, 직교하면 0, 정반대면 -1입니다. (아래 코드에서는 1에서 이 값을 뺀 코사인 *거리*를 사용하므로 값이 작을수록 가깝습니다.)

$$\text{similarity}(\mathbf{A}, \mathbf{B}) = \cos(\theta) = \frac{\mathbf{A} \cdot \mathbf{B}}{||\mathbf{A}|| \cdot ||\mathbf{B}||} = \frac{\sum_{i=1}^{n} a_i b_i}{\sqrt{\sum_{i=1}^{n} a_i^2} \sqrt{\sum_{i=1}^{n} b_i^2}}$$

#### 내적 (스칼라 곱)

내적(스칼라 곱)은 한 벡터를 다른 벡터에 투영한 정도를 나타냅니다. 특히 벡터가 이미 정규화되어 있는 경우(L2 노름이 1인 경우) 유사도 척도로 사용되며, 이때는 코사인 유사도와 같아집니다. 값이 클수록 가깝습니다(좋습니다).

$$\text{similarity}(\mathbf{A}, \mathbf{B}) = \mathbf{A} \cdot \mathbf{B} = \sum_{i=1}^{n} a_i b_i$$

```python
import numpy as np

def vector_distance(vec1, vec2, distance_type='l2'):
    """
    여러 가지 척도로 두 벡터 사이의 거리 또는 내적을 계산합니다.

    Args:
        vec1 (numpy.ndarray): 첫 번째 벡터.
        vec2 (numpy.ndarray): 두 번째 벡터.
        distance_type (str, optional): 수행할 계산의 종류.
                                      'l1', 'l2', 'cosine', 'euclidean', 'dot' 중 하나.
                                      기본값은 'l2'.

    Returns:
        float: 계산된 거리 또는 내적 값.

    Raises:
        ValueError: distance_type이 올바르지 않은 경우.
        ValueError: 입력 벡터의 shape이 서로 다른 경우('dot' 제외).
    """

    vec1 = np.array(vec1)
    vec2 = np.array(vec2)

    if distance_type != 'dot' and vec1.shape != vec2.shape:
        raise ValueError("거리 계산을 하려면 두 입력 벡터의 shape이 같아야 합니다.")

    if distance_type == 'l1':
        return np.sum(np.abs(vec1 - vec2))
    elif distance_type == 'l2' or distance_type == 'euclidean':
        return np.linalg.norm(vec1 - vec2)
    elif distance_type == 'cosine':
        dot_product = np.dot(vec1, vec2)
        norm_vec1 = np.linalg.norm(vec1)
        norm_vec2 = np.linalg.norm(vec2)
        if norm_vec1 == 0 or norm_vec2 == 0:
            return np.nan  # 영벡터 처리.
        return 1 - (dot_product / (norm_vec1 * norm_vec2))
    elif distance_type == 'dot':
        return np.dot(vec1, vec2)
    else:
        raise ValueError("distance_type이 올바르지 않습니다. 'l1', 'l2', 'cosine', 'euclidean', 'dot' 중에서 선택하세요.")

def vec_sim(vec_a,vec_b):
  return {
    "l1_dist": vector_distance(vec_a, vec_b, 'l1'), # 맨해튼 거리, 작을수록 좋음
    "l2_dist": vector_distance(vec_a, vec_b, 'l2'), # 유클리드 거리, 작을수록 좋음 (다차원 피타고라스)
    "cosine_dist": vector_distance(vec_a, vec_b, 'cosine'), # 코사인 거리, 작을수록 좋음
    "dot_product": vector_distance(vec_a, vec_b, 'dot') # 내적, 클수록 좋음 (상관도가 높음)
  }
```
